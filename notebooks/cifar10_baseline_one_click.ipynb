# CIFAR-10 FedAvg baseline: one click

Runtime → Change runtime type → **T4 GPU**, then Runtime → **Run all**.
It clones the `fl-baseline` branch, runs the tests, runs the CIFAR-10 baseline (50 clients, 10 per round, Dirichlet α=0.1, 150 rounds), and downloads `results_cifar10.zip` at the end.
Expect about 10–15 minutes on a T4 (an estimate). CIFAR-10 (~170 MB) downloads on the first run.

```
# 1. Get the code (fl-baseline branch) and install requirements. Colab already has torch.
import os
REPO = "/content/Carbon-Aware-Federated-Learning"
if not os.path.exists(REPO):
    !git clone -q -b fl-baseline https://github.com/Harshil089/Carbon-Aware-Federated-Learning.git {REPO}
else:
    !git -C {REPO} pull -q
%cd {REPO}
!pip install -q -r requirements.txt
import torch
print("torch", torch.__version__, "| CUDA available:", torch.cuda.is_available())
```

```
# 2. Unit tests (FedAvg, partitions, energy, simulated latency/energy/carbon)
!python -m pytest -q tests
```

```
# 3. CIFAR-10 baseline
!python experiments/baseline/run_baseline.py --config configs/experiment_cifar10.yaml --alpha 0.1
```

```
# 4. Show the summary, check which energy sources were used, and download the results
import glob, os
from google.colab import files
run = max(glob.glob("results/*cifar10*"), key=os.path.getmtime)
print(open(f"{run}/summary.json").read())
print(open(f"{run}/energy.json").read())   # gpu_source should read nvml_measured on a T4
!zip -qr results_cifar10.zip {run}
files.download("results_cifar10.zip")
```
